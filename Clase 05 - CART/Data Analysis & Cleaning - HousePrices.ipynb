## Antes de tocar nada: tipos y duplicados

Siguiendo el checklist de profiling, lo primero es entender qué es cada columna y revisar si hay filas repetidas, antes de meternos con los faltantes.

### Filas duplicadas

Un chequeo rápido: ¿hay filas exactamente repetidas? En este dataset cada fila es una casa con su `Id` único, así que no deberíamos encontrar ninguna, pero conviene confirmarlo siempre.

```python
n_dup = data.duplicated().sum()
print("Filas duplicadas (fila completa):", n_dup)
# Si sacamos el Id, que es un identificador unico, tampoco deberia haber casas identicas en el resto
n_dup_sin_id = data.drop(columns='Id').duplicated().sum()
print("Duplicadas ignorando el Id:", n_dup_sin_id)

# Si hubiera duplicados exactos, los eliminamos quedandonos con la primera aparicion
if n_dup > 0:
    data = data.drop_duplicates().reset_index(drop=True)
    print("Se eliminaron", n_dup, "filas duplicadas. Filas restantes:", len(data))
else:
    print("No hay duplicados para eliminar.")
```

Vemos que correlaciona bastante alto (0.35) con SalePrice, por lo tanto es una variable bastante importante para después entrenar. Siendo tantos no podemos estar borrandolos. Tenemos varias opciones:

Opción 1: Llenar con la media. ¿Está bien? ¿Está mal? Hay que pensarlo mucho. A mi me parece que llenar con la media esta mal porque tenemos muchas casas de distintos tipos y formas y vamos a distorsionar mucho la realidad.

Opción 2: Llenar con la media pero por vecindario. Mejor que llenar con la media, pero capaz incompleto de todos modos. En este caso lo que haríamos es hacer un groupby por vecindario, calcular la media, y ponerle eso. ¿Se entiende por qué es bastante más acertado que simplemente la media? De todos modos, corremos riesgos.

Opción 3: Hacer una regresión y llenar con eso. Bastante más acertado, aunque también es un paso más que tenemos que hacer y, dependiendo de la complejidad, puede salir bien o mal.

Cotejemos entonces solamente las opciones 2 y 3.

_Media por Neighborhood_

Pros: Es simple, es directo y fácil de implementar. Solo necesitamos calcular la mediana de 'LotFrontage' para cada vecindario y aplicar cada valor a las propiedades que le falta ese dato. Dado que los vecindarios pueden tener características parecidas y compartidas es ampliamente superior hacer esto a una mediana global de todas las casas.

Contras: Perdemos información de variabilidad dentro de un vecindario. Vamos a tener casas parecidas en ese aspecto y va a suavizar las diferencias entre propiedades del mismo barrio. También, estamos asumiendo homogeneidad en el vecindario, lo cual es imposible de chequear o realmente saber si estamos haciendolo bien.

_Regresión_

Pros: Mayor precisión, muchísima mayor precisión. Podemos usar distintas variables como LotArea, 1stFlrSF y demás, lo cual lleva a una imputación mucho más precisa y específica para cada propiedad. Vamos a manejar también mucho mejor los outliers y casos especiales que tirando la media.

Contras: Requiere tiempo y esfuerzo. Vamos a tener que construir, ajustar y validar el modelo antes de aplicarlo para la imputación. Corremos también el riesgo de sobreajustar un montón.

**Conclusión**

Si tenemos tiempo y ganas, hacemos una regresión. Si necesitamos responder rápido y simple, imputamos por vecindario. ¿Qué hacemos?

```python
# Imputamos con la MEDIANA de LotFrontage de cada vecindario (mas robusta que la media
# ante casas atipicas dentro del barrio). Si algun barrio quedara sin ningun dato, usamos
# la mediana global como respaldo.
data['LotFrontage'] = data.groupby('Neighborhood')['LotFrontage'].transform(
    lambda x: x.fillna(x.median())
)
data['LotFrontage'] = data['LotFrontage'].fillna(data['LotFrontage'].median())

print("Nulos en LotFrontage despues de imputar:", data['LotFrontage'].isnull().sum())
```

## Detección de outliers

Ahora que no quedan faltantes, podemos mirar las distribuciones de las variables numéricas y buscar valores extremos. Nos concentramos en las que más importan para el precio: `SalePrice` (la target) y `GrLivArea` (superficie habitable), que es la más correlacionada con el precio.

### La regla del rango intercuartil (IQR)

Un valor se considera outlier si cae fuera del intervalo [Q1 − 1.5·IQR, Q3 + 1.5·IQR]. Lo aplicamos a `GrLivArea`:

```python
def limites_iqr(serie):
    q1 = serie.quantile(0.25)
    q3 = serie.quantile(0.75)
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr

low, high = limites_iqr(data['GrLivArea'])
print(f"Limites GrLivArea: [{low:.0f}, {high:.0f}]")
outliers = data[(data['GrLivArea'] < low) | (data['GrLivArea'] > high)]
print(f"Casas fuera del rango: {len(outliers)}")
```

### Marcar un outlier no es lo mismo que borrarlo

La regla del IQR nos marca candidatos, pero la decisión es nuestra. Miremos el caso más conocido de este dataset: las casas con muchísima superficie que se vendieron muy baratas.

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='GrLivArea', y='SalePrice', data=data, color='#6e8f6a')
plt.title('Superficie vs precio')
plt.show()

# Las dos casas enormes (>4000 sqft) y baratas rompen la relacion superficie-precio
raras = data[(data['GrLivArea'] > 4000) & (data['SalePrice'] < 300000)]
print(raras[['GrLivArea', 'SalePrice', 'OverallQual', 'SaleCondition']])
```

Estas dos casas grandes y baratas son un caso real, no un error de carga: mirando `SaleCondition` se ve que fueron ventas atípicas (`Partial`, ventas de obra en construcción). No las borramos en silencio: la decisión de excluirlas depende de la pregunta. Para un modelo del mercado minorista normal, tiene sentido sacarlas y **dejarlo documentado**; para un análisis de todas las ventas, no.

Acá las dejamos, pero anotamos que existen. Ese es el punto: el profiling detecta y decide, no borra automáticamente.

## Codificación de variables categóricas

El último paso antes de poder modelar es transformar las categóricas en números, porque la mayoría de los algoritmos no trabajan con texto. La técnica más común es **one-hot encoding**: cada categoría se convierte en una columna binaria (0/1).

Ojo: no estandarizamos las variables numéricas. Los árboles (CART) parten por umbrales y no miden distancias, así que la escala no les afecta. La estandarización recién haría falta para modelos basados en distancia (como KNN), y en ese caso se hace *después* de separar train y test, para no filtrar información del test al entrenamiento.

```python
# Verificamos que no queden faltantes antes de codificar
faltantes = data.isnull().sum().sum()
print("Faltantes totales en el dataset:", faltantes)
assert faltantes == 0, "Todavia quedan faltantes: hay que resolverlos antes de codificar"

categoricas = data.select_dtypes(include=['object', 'category']).columns
print(f"Columnas categoricas a codificar: {len(categoricas)}")
print(f"Dimensiones antes del encoding: {data.shape}")
```

```python
data_encoded = pd.get_dummies(data, columns=categoricas, drop_first=True)
print(f"Dimensiones despues del encoding: {data_encoded.shape}")
data_encoded.head()
```

Pasamos de ~80 columnas a varios cientos, porque cada categoría generó su propia columna binaria. Con `drop_first=True` eliminamos una categoría de cada variable (la que queda representada cuando todas las demás son 0), para evitar redundancia.

Con esto el dataset queda **listo para modelar**: sin faltantes, con los tipos corregidos, los outliers identificados y las categóricas codificadas. El paso de entrenar, como decíamos al principio, es el más fácil.
