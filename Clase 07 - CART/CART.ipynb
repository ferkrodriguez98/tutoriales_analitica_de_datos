# CART

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree

data = {
    'Departamento': ['Ventas', 'Soporte', 'Logística', 'Ventas', 'Soporte', 'Logística'],
    'Antigüedad': [2, 5, 3, 1, 4, 6],
    'Cumplimiento de Objetivos': [80, 60, 75, 50, 85, 70],
    'Recibe Bono': ['No', 'No', 'Sí', 'No', 'Sí', 'No']
}

df = pd.DataFrame(data)
df['Recibe Bono'] = df['Recibe Bono'].map({'No':0, 'Sí':1})

# Codificar variable categórica
df_encoded = pd.get_dummies(df, columns=['Departamento'])

# Variables predictoras
X = df_encoded.drop(columns=['Recibe Bono'])
y = df_encoded['Recibe Bono']

# Entrenar árbol
cart_model = DecisionTreeClassifier(random_state=42)
cart_model.fit(X, y)

# Visualizar árbol
plt.figure(figsize=(12, 8))
plot_tree(cart_model, feature_names=X.columns, class_names=['No', 'Sí'], filled=True, rounded=True)
plt.show()
```

**Nota**: El Gini que dice en la caja no es el Gini en la división.
